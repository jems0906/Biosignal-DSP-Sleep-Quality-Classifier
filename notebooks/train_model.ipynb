# Train a subject-level Sleep-EDF model

This notebook runs the same reproducible training workflow used by the API: download a manageable Sleep-EDF subset with MNE, curate 30-second EEG epochs, extract normalized delta/theta/alpha/beta bandpower, split by subject, train a RandomForest, and save the model plus validation artifacts.

```python
from pathlib import Path
import json
import subprocess
import sys

ROOT = Path.cwd()
if not (ROOT / "backend").exists():
    ROOT = ROOT.parent
TRAINER = ROOT / "backend" / "scripts" / "train_model.py"
SUBJECTS = [0, 1, 2, 3, 4]
print(f"Training from {TRAINER} with subjects {SUBJECTS}")
```

## Download, curate, split, and train

The trainer keeps all epochs from a subject in exactly one partition. It writes `backend/models/model.pkl` and `backend/models/metrics.json`; the API automatically loads those files on the next request.

```python
result = subprocess.run([sys.executable, str(TRAINER), "--subjects", *[str(subject) for subject in SUBJECTS]], cwd=ROOT, check=True, capture_output=True, text=True)
print(result.stdout[-4000:])
metrics = json.loads((ROOT / "backend" / "models" / "metrics.json").read_text(encoding="utf-8"))
print({key: metrics[key] for key in ("balanced_accuracy", "macro_f1", "split")})
```
